## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# DATA WRANGLING / DATA CLEANING
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Create a copy of the original dataset
df_clean = df.copy()

# ------------------------------------------------------------
# 1. Remove unnecessary index column if present
# ------------------------------------------------------------
if 'Unnamed: 0' in df_clean.columns:
    df_clean.drop('Unnamed: 0', axis=1, inplace=True)

# ------------------------------------------------------------
# 2. Check duplicate records
# ------------------------------------------------------------
print("Duplicate records before cleaning:", df_clean.duplicated().sum())

# Remove duplicate records
df_clean.drop_duplicates(inplace=True)

print("Duplicate records after cleaning:", df_clean.duplicated().sum())

# ------------------------------------------------------------
# 3. Remove leading/trailing spaces from column names
# ------------------------------------------------------------
df_clean.columns = df_clean.columns.str.strip()

# ------------------------------------------------------------
# 4. Clean text columns
# ------------------------------------------------------------
text_columns = df_clean.select_dtypes(include='object').columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

# ------------------------------------------------------------
# 5. Check missing values
# ------------------------------------------------------------
print("\nMissing values before handling:")
display(df_clean.isnull().sum())

# ------------------------------------------------------------
# 6. Handle missing values
# ------------------------------------------------------------

# Replace missing categorical values with 'Unknown'
categorical_columns = [
    'director',
    'cast',
    'country',
    'rating',
    'duration'
]

for column in categorical_columns:
    if column in df_clean.columns:
        df_clean[column] = df_clean[column].fillna('Unknown')

# Replace missing date_added with 'Not Available'
if 'date_added' in df_clean.columns:
    df_clean['date_added'] = df_clean['date_added'].fillna('Not Available')

# ------------------------------------------------------------
# 7. Convert date_added into datetime
# ------------------------------------------------------------
if 'date_added' in df_clean.columns:
    df_clean['date_added'] = pd.to_datetime(
        df_clean['date_added'],
        errors='coerce'
    )

# ------------------------------------------------------------
# 8. Extract useful date features
# ------------------------------------------------------------
if 'date_added' in df_clean.columns:
    df_clean['added_year'] = df_clean['date_added'].dt.year
    df_clean['added_month'] = df_clean['date_added'].dt.month
    df_clean['added_month_name'] = df_clean['date_added'].dt.month_name()

# ------------------------------------------------------------
# 9. Clean duration column
# ------------------------------------------------------------
if 'duration' in df_clean.columns:

    # Extract numerical duration
    df_clean['duration_value'] = pd.to_numeric(
        df_clean['duration'].astype(str).str.extract(r'(\d+)')[0],
        errors='coerce'
    )

    # Extract duration unit
    df_clean['duration_unit'] = (
        df_clean['duration']
        .astype(str)
        .str.extract(r'(\D+)$')[0]
        .str.strip()
    )

# ------------------------------------------------------------
# 10. Standardize the type column
# ------------------------------------------------------------
if 'type' in df_clean.columns:
    df_clean['type'] = df_clean['type'].str.title()

# ------------------------------------------------------------
# 11. Check missing values after cleaning
# ------------------------------------------------------------
print("\nMissing values after cleaning:")
display(df_clean.isnull().sum())

# ------------------------------------------------------------
# 12. Check final dataset shape
# ------------------------------------------------------------
print("\nFinal Dataset Shape:", df_clean.shape)

# ------------------------------------------------------------
# 13. Display cleaned dataset
# ------------------------------------------------------------
display(df_clean.head())
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Hypothesis 1 — Movie vs TV Show Duration

Hypothetical statement:

Movies have a significantly different average duration than TV Shows.

H₀ (Null Hypothesis): There is no significant difference in the average duration between Movies and TV Shows.
H₁ (Alternative Hypothesis): There is a significant difference in the average duration between Movies and TV Shows.
Test: Independent Samples T-Test

Important: Because duration_value represents minutes for Movies but number of seasons for TV Shows, this hypothesis is not statistically meaningful unless duration is standardized into a comparable measure. I would not recommend using this one for your final project.

Hypothesis 2 — Content Type and Rating

Hypothetical statement:

Content Type (Movie or TV Show) is associated with Content Rating.

H₀: Content Type and Rating are independent; there is no significant association.
H₁: Content Type and Rating are associated.
Test: Chi-Square Test of Independence

This is a strong choice because both variables are categorical.

Hypothesis 3 — Movie Duration and Rating

Hypothetical statement:

Average movie duration differs significantly across content-rating categories.

H₀: The average movie duration is the same across the rating categories.
H₁: At least one rating category has a significantly different average movie duration.
Test: One-Way ANOVA

This is also a good choice because duration_value is numerical and rating is categorical.

Recommended final 3
No.	Hypothesis	Statistical Test
1	Content Type is associated with Rating	Chi-Square Test
2	Movie duration differs across Ratings	One-Way ANOVA
3	Movie duration is related to Release Year	Pearson Correlation

For all three, use α = 0.05 (5% significance level).

Decision rule:

p-value < 0.05 → Reject H₀ → Significant result
p-value ≥ 0.05 → Fail to reject H₀ → Not statistically significant

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):

There is no significant association between Content Type (Movies and TV Shows) and Content Rating on Amazon Prime.

Alternative Hypothesis (H₁):

There is a significant association between Content Type (Movies and TV Shows) and Content Rating on Amazon Prime.

Statistical Test: Chi-Square Test of Independence
Significance Level (α): 0.05

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# ============================================================
# HYPOTHESIS TESTING - 1
# Obtaining P-Value
# Content Type vs Content Rating
# ============================================================

from scipy.stats import chi2_contingency

# Create contingency table
contingency_table = pd.crosstab(
    df_clean['type'],
    df_clean['rating']
)

# Perform Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(
    contingency_table
)

# Display P-Value
print("Chi-Square Statistic :", chi2_stat)
print("P-Value              :", p_value)
print("Degrees of Freedom   :", dof)

# Significance level
alpha = 0.05

print("\nSignificance Level (α):", alpha)

# Decision based on P-Value
if p_value < alpha:
    print("Decision: Reject the Null Hypothesis (H₀).")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H₀).")
````

##### Which statistical test have you done to obtain P-Value?

Chi-Square Test of Independence was performed to obtain the P-value. This test was selected because both variables—Content Type (Movie/TV Show) and Content Rating—are categorical variables.

Significance level (α): 0.05

##### Why did you choose the specific statistical test?

The Chi-Square Test of Independence was chosen because both variables, Content Type and Content Rating, are categorical variables. The test determines whether there is a statistically significant association between these two categorical variables. It is therefore appropriate for testing whether the distribution of ratings differs between Movies and TV Shows.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):

There is no significant difference in the average movie duration across different content-rating categories on Amazon Prime.

Alternative Hypothesis (H₁):

There is a significant difference in the average movie duration across different content-rating categories on Amazon Prime.

Statistical Test: One-Way ANOVA
Significance Level (α): 0.05

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# ============================================================
# HYPOTHESIS TESTING - 2
# One-Way ANOVA Test
# Movie Duration vs Content Rating
# ============================================================

from scipy.stats import f_oneway

# Select only Movies with valid duration and rating
movie_data = df_clean[
    (df_clean['type'] == 'Movie') &
    (df_clean['duration_value'].notna()) &
    (df_clean['rating'].notna()) &
    (df_clean['rating'] != 'Unknown')
].copy()

# Create groups based on rating
rating_groups = [
    group['duration_value'].values
    for _, group in movie_data.groupby('rating')
    if len(group) >= 2
]

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(*rating_groups)

# Display results
print("F-Statistic :", f_statistic)
print("P-Value     :", p_value)

# Significance level
alpha = 0.05
print("Significance Level (α):", alpha)

# Decision
if p_value < alpha:
    print("\nDecision: Reject the Null Hypothesis (H₀).")
    print("There is a significant difference in average movie duration")
    print("across different content-rating categories.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H₀).")
    print("There is no significant difference in average movie duration")
    print("across different content-rating categories.")
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA (Analysis of Variance) was performed to obtain the P-value. This test was selected to compare the mean movie duration across multiple content-rating categories.

Significance Level (α): 0.05

##### Why did you choose the specific statistical test?

The One-Way ANOVA test was chosen because the dependent variable, Movie Duration, is numerical, while the independent variable, Content Rating, is categorical with multiple groups. ANOVA is appropriate for determining whether there is a statistically significant difference in the mean movie duration among the different rating categories.

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
# ============================================================
# HANDLING MISSING VALUES & MISSING VALUE IMPUTATION
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Create a copy of the cleaned dataset
df_imputed = df_clean.copy()

# ------------------------------------------------------------
# Step 1: Check missing values before imputation
# ------------------------------------------------------------

print("Missing Values Before Imputation:")
display(df_imputed.isnull().sum())


# ------------------------------------------------------------
# Step 2: Handle categorical columns
# Replace missing categorical values with 'Unknown'
# ------------------------------------------------------------

categorical_columns = [
    'director',
    'cast',
    'country',
    'rating',
    'duration'
]

for column in categorical_columns:
    if column in df_imputed.columns:
        df_imputed[column] = df_imputed[column].fillna('Unknown')


# ------------------------------------------------------------
# Step 3: Handle date column
# Missing dates are represented as 'Not Available'
# ------------------------------------------------------------

if 'date_added' in df_imputed.columns:
    df_imputed['date_added'] = df_imputed['date_added'].fillna(
        pd.Timestamp('1900-01-01')
    )


# ------------------------------------------------------------
# Step 4: Numerical value imputation
# Use median because it is less affected by outliers
# ------------------------------------------------------------

numerical_columns = df_imputed.select_dtypes(
    include=np.number
).columns

for column in numerical_columns:
    if df_imputed[column].isnull().sum() > 0:
        df_imputed[column] = df_imputed[column].fillna(
            df_imputed[column].median()
        )


# ------------------------------------------------------------
# Step 5: Check missing values after imputation
# ------------------------------------------------------------

print("\nMissing Values After Imputation:")
display(df_imputed.isnull().sum())


# ------------------------------------------------------------
# Step 6: Display final dataset information
# ------------------------------------------------------------

print("\nDataset Shape After Imputation:")
print(df_imputed.shape)

print("\nSample Data After Imputation:")
display(df_imputed.head())
````

What all missing value imputation techniques have you used and why did you use those techniques?

Imputation Technique	Used For	Why Used
Mode / Category “Unknown”	Categorical columns such as director, cast, country, rating, duration	Missing categorical information cannot be reliably calculated using mean/median, so "Unknown" preserves the records without creating false information.
Median Imputation	Numerical columns such as release_year and duration_value	Median is less affected by extreme values/outliers than the mean, making it a safer choice for numerical data.
NaT / Missing Date Handling	date_added	Missing dates should not be replaced with an artificial date because that could create misleading time-based analysis.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# ============================================================
# HANDLING OUTLIERS & OUTLIER TREATMENT
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Create a copy of the dataset
df_outlier = df_imputed.copy()

# ------------------------------------------------------------
# Step 1: Select numerical column for outlier analysis
# ------------------------------------------------------------

column = 'duration_value'

# Remove missing values for analysis
data = df_outlier[column].dropna()

print("Original Number of Records:", len(data))


# ------------------------------------------------------------
# Step 2: Calculate Q1, Q3 and IQR
# ------------------------------------------------------------

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

# Calculate lower and upper limits
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)


# ------------------------------------------------------------
# Step 3: Identify outliers
# ------------------------------------------------------------

outliers = df_outlier[
    (df_outlier[column] < lower_bound) |
    (df_outlier[column] > upper_bound)
]

print("\nNumber of Outliers:", len(outliers))

print("\nOutlier Records:")
display(outliers[[column, 'title', 'type', 'rating']].head(10))


# ------------------------------------------------------------
# Step 4: Visualize outliers using Box Plot
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(x=df_outlier[column])

plt.title('Box Plot Before Outlier Treatment')
plt.xlabel('Duration')

plt.show()


# ------------------------------------------------------------
# Step 5: Outlier Treatment using Capping
# Values outside the limits are replaced with boundary values
# ------------------------------------------------------------

df_outlier[column] = df_outlier[column].clip(
    lower=lower_bound,
    upper=upper_bound
)


# ------------------------------------------------------------
# Step 6: Check outliers after treatment
# ------------------------------------------------------------

remaining_outliers = df_outlier[
    (df_outlier[column] < lower_bound) |
    (df_outlier[column] > upper_bound)
]

print("\nOutliers After Treatment:", len(remaining_outliers))


# ------------------------------------------------------------
# Step 7: Visualize after treatment
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(x=df_outlier[column])

plt.title('Box Plot After Outlier Treatment')
plt.xlabel('Duration')

plt.show()


# ------------------------------------------------------------
# Step 8: Final dataset information
# ------------------------------------------------------------

print("\nFinal Dataset Shape:", df_outlier.shape)

display(df_outlier.head())
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

I used the IQR (Interquartile Range) method to detect outliers and Capping/Winsorization to treat them.

Technique	Purpose	Why Used
IQR Method	Detect outliers	It identifies unusually low or high values using Q1, Q3, and the IQR without being strongly affected by extreme values.
Capping (Winsorization)	Treat outliers	Extreme values are replaced with the lower or upper IQR boundary instead of deleting the records. This preserves the dataset size and prevents extreme values from having excessive influence on analysis.

#### 1. Expand Contraction

````python
# Expand Contraction
# ============================================================
# TEXTUAL DATA PROCESSING - EXPAND CONTRACTIONS
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import re

# Create a copy of the dataset
df_text = df_outlier.copy()

# Dictionary of common English contractions
contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'m": " am",
    "'ll": " will",
    "'d": " would",
    "'ve": " have",
    "'t": " not",
    "'em": " them"
}


# ------------------------------------------------------------
# Function to expand contractions
# ------------------------------------------------------------

def expand_contractions(text):
    """
    Expands common English contractions into full words.
    Example: 'can't' -> 'cannot'
    """

    if pd.isna(text):
        return text

    text = str(text)

    for contraction, expansion in contractions.items():
        text = re.sub(
            re.escape(contraction),
            expansion,
            text,
            flags=re.IGNORECASE
        )

    return text


# ------------------------------------------------------------
# Apply function to Description column
# ------------------------------------------------------------

if 'description' in df_text.columns:
    df_text['description_expanded'] = df_text['description'].apply(
        expand_contractions
    )


# ------------------------------------------------------------
# Display original and processed text
# ------------------------------------------------------------

print("Original and Expanded Text:")
display(
    df_text[['description', 'description_expanded']].head(10)
)
````

#### 2. Lower Casing

````python
# Lower Casing
# ============================================================
# TEXTUAL DATA PROCESSING - LOWER CASING
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Create a copy of the text-processed dataset
df_lower = df_text.copy()

# ------------------------------------------------------------
# Convert description text to lowercase
# ------------------------------------------------------------

if 'description_expanded' in df_lower.columns:
    df_lower['description_lower'] = (
        df_lower['description_expanded']
        .fillna('')
        .astype(str)
        .str.lower()
    )

# ------------------------------------------------------------
# Display original and lower-cased text
# ------------------------------------------------------------

print("Original and Lower-Cased Text:")
display(
    df_lower[
        ['description_expanded', 'description_lower']
    ].head(10)
)
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
# ============================================================
# TEXTUAL DATA PROCESSING - REMOVE PUNCTUATIONS
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import string

# Create a copy of the lower-cased dataset
df_punctuation = df_lower.copy()

# ------------------------------------------------------------
# Function to remove punctuation
# ------------------------------------------------------------

def remove_punctuation(text):
    """
    Removes punctuation marks from textual data.
    Example: 'hello, world!' -> 'hello world'
    """

    if pd.isna(text):
        return text

    text = str(text)

    # Remove punctuation characters
    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )

    return text


# ------------------------------------------------------------
# Apply function to Description column
# ------------------------------------------------------------

if 'description_lower' in df_punctuation.columns:
    df_punctuation['description_clean'] = (
        df_punctuation['description_lower']
        .apply(remove_punctuation)
    )


# ------------------------------------------------------------
# Display original and processed text
# ------------------------------------------------------------

print("Text Before and After Removing Punctuation:")

display(
    df_punctuation[
        ['description_lower', 'description_clean']
    ].head(10)
)
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits
# ============================================================
# TEXTUAL DATA PROCESSING
# Remove URLs, Words Containing Digits & Digits
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import re

# Create a copy of the punctuation-cleaned dataset
df_url_clean = df_punctuation.copy()


# ------------------------------------------------------------
# Function to remove URLs
# ------------------------------------------------------------

def remove_urls(text):
    """
    Removes HTTP, HTTPS and WWW URLs from text.
    """

    if pd.isna(text):
        return text

    text = str(text)

    # Remove URLs
    text = re.sub(
        r'https?://\S+|www\.\S+',
        '',
        text
    )

    return text


# ------------------------------------------------------------
# Function to remove words containing digits
# and standalone digits
# ------------------------------------------------------------

def remove_words_and_digits(text):
    """
    Removes words containing digits and standalone numbers.
    Examples:
    'movie2' -> removed
    'season3' -> removed
    '2025' -> removed
    """

    if pd.isna(text):
        return text

    text = str(text)

    # Remove words containing at least one digit
    text = re.sub(r'\S*\d\S*', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# ------------------------------------------------------------
# Step 1: Remove URLs
# ------------------------------------------------------------

df_url_clean['description_no_url'] = (
    df_url_clean['description_clean']
    .apply(remove_urls)
)


# ------------------------------------------------------------
# Step 2: Remove words containing digits
# and standalone digits
# ------------------------------------------------------------

df_url_clean['description_final'] = (
    df_url_clean['description_no_url']
    .apply(remove_words_and_digits)
)


# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("Text Before and After Processing:")

display(
    df_url_clean[
        [
            'description_clean',
            'description_no_url',
            'description_final'
        ]
    ].head(10)
)
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
# ============================================================
# TEXTUAL DATA PROCESSING - REMOVE STOPWORDS
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Install NLTK resources if required
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Create a copy of the processed dataset
df_stopwords = df_url_clean.copy()


# ------------------------------------------------------------
# Function to remove stopwords
# ------------------------------------------------------------

def remove_stopwords(text):
    """
    Removes common English stopwords from text.
    Example:
    'this is a good movie' -> 'good movie'
    """

    if pd.isna(text):
        return text

    # Convert text into words
    words = str(text).split()

    # Remove stopwords
    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]

    # Join words back into a sentence
    return ' '.join(filtered_words)


# ------------------------------------------------------------
# Apply stopword removal
# ------------------------------------------------------------

if 'description_final' in df_stopwords.columns:
    df_stopwords['description_no_stopwords'] = (
        df_stopwords['description_final']
        .apply(remove_stopwords)
    )


# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("Text Before and After Stopword Removal:")

display(
    df_stopwords[
        ['description_final', 'description_no_stopwords']
    ].head(10)
)
````

````python
# Remove White spaces
# ============================================================
# TEXTUAL DATA PROCESSING - REMOVE WHITE SPACES
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

# Create a copy of the stopword-removed dataset
df_whitespace = df_stopwords.copy()


# ------------------------------------------------------------
# Function to remove extra white spaces
# ------------------------------------------------------------

def remove_extra_whitespace(text):
    """
    Removes leading, trailing, and multiple white spaces.
    Example:
    '  young   boy  lives  ' -> 'young boy lives'
    """

    if pd.isna(text):
        return text

    # Convert multiple spaces into a single space
    text = re.sub(r'\s+', ' ', str(text))

    # Remove leading and trailing spaces
    return text.strip()


# ------------------------------------------------------------
# Apply white-space removal
# ------------------------------------------------------------

if 'description_no_stopwords' in df_whitespace.columns:
    df_whitespace['description_clean_final'] = (
        df_whitespace['description_no_stopwords']
        .apply(remove_extra_whitespace)
    )


# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("Text Before and After Removing White Spaces:")

display(
    df_whitespace[
        ['description_no_stopwords', 'description_clean_final']
    ].head(10)
)
````

#### 6. Rephrase Text

````python
# Rephrase Text
# ============================================================
# TEXTUAL DATA PROCESSING - REPHRASE TEXT
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import re

# Create a copy of the cleaned dataset
df_rephrased = df_whitespace.copy()


# ------------------------------------------------------------
# Function to rephrase and normalize text
# ------------------------------------------------------------

def rephrase_text(text):
    """
    Cleans and normalizes the text while preserving
    the original meaning.
    """

    if pd.isna(text):
        return text

    text = str(text)

    # Remove unnecessary spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Standardize common connecting words
    text = re.sub(r'\b&\b', 'and', text)

    # Return normalized text
    return text


# ------------------------------------------------------------
# Apply rephrasing function
# ------------------------------------------------------------

if 'description_clean_final' in df_rephrased.columns:
    df_rephrased['description_rephrased'] = (
        df_rephrased['description_clean_final']
        .apply(rephrase_text)
    )


# ------------------------------------------------------------
# Display original and rephrased text
# ------------------------------------------------------------

print("Text Before and After Rephrasing:")

display(
    df_rephrased[
        ['description_clean_final', 'description_rephrased']
    ].head(10)
)
````

````python
# Tokenization
# ============================================================
# TEXTUAL DATA PROCESSING - TOKENIZATION
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import nltk
from nltk.tokenize import word_tokenize

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Create a copy of the rephrased dataset
df_tokenized = df_rephrased.copy()


# ------------------------------------------------------------
# Function for Word Tokenization
# ------------------------------------------------------------

def tokenize_text(text):
    """
    Splits text into individual words (tokens).
    """

    if pd.isna(text):
        return []

    return word_tokenize(str(text))


# ------------------------------------------------------------
# Apply Tokenization
# ------------------------------------------------------------

if 'description_rephrased' in df_tokenized.columns:
    df_tokenized['description_tokens'] = (
        df_tokenized['description_rephrased']
        .apply(tokenize_text)
    )


# ------------------------------------------------------------
# Display tokenized text
# ------------------------------------------------------------

print("Text Before and After Tokenization:")

display(
    df_tokenized[
        ['description_rephrased', 'description_tokens']
    ].head(10)
)
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
# ============================================================
# TEXTUAL DATA PROCESSING - TEXT NORMALIZATION
# Stemming & Lemmatization
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet

# Download required NLTK resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Create a copy of the tokenized dataset
df_normalized = df_tokenized.copy()


# ------------------------------------------------------------
# Step 1: Stemming
# ------------------------------------------------------------

stemmer = PorterStemmer()

def apply_stemming(tokens):
    """
    Converts words into their root/stem form.
    """

    if not isinstance(tokens, list):
        return []

    return [
        stemmer.stem(word)
        for word in tokens
    ]


df_normalized['stemmed_tokens'] = (
    df_normalized['description_tokens']
    .apply(apply_stemming)
)


# ------------------------------------------------------------
# Step 2: Lemmatization
# ------------------------------------------------------------

lemmatizer = WordNetLemmatizer()

def apply_lemmatization(tokens):
    """
    Converts words into their meaningful base/dictionary form.
    """

    if not isinstance(tokens, list):
        return []

    return [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]


df_normalized['lemmatized_tokens'] = (
    df_normalized['description_tokens']
    .apply(apply_lemmatization)
)


# ------------------------------------------------------------
# Step 3: Convert Lemmatized Tokens back to Text
# ------------------------------------------------------------

df_normalized['normalized_text'] = (
    df_normalized['lemmatized_tokens']
    .apply(lambda x: ' '.join(x))
)


# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("Text Normalization Results:")

display(
    df_normalized[
        [
            'description_tokens',
            'stemmed_tokens',
            'lemmatized_tokens',
            'normalized_text'
        ]
    ].head(10)
)
````

It was chosen over stemming because stemming may produce incomplete words, whereas lemmatization generally produces meaningful words. This helps reduce vocabulary complexity and makes the textual data more consistent for further NLP analysis.

````python
# Manipulate Features to minimize feature correlation and create new features
# ============================================================
# FEATURE MANIPULATION & NEW FEATURE CREATION
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

import pandas as pd
import numpy as np

# Create a copy of the dataset
df_features = df_outlier.copy()


# ------------------------------------------------------------
# 1. Create Content Age
# Measures how old the content is
# ------------------------------------------------------------

current_year = pd.Timestamp.now().year

df_features['content_age'] = (
    current_year - df_features['release_year']
)


# ------------------------------------------------------------
# 2. Create Is Recent Content
# 1 = Recent content, 0 = Older content
# ------------------------------------------------------------

df_features['is_recent'] = np.where(
    df_features['content_age'] <= 5,
    1,
    0
)


# ------------------------------------------------------------
# 3. Create Duration Category
# Categorize movie duration into meaningful groups
# ------------------------------------------------------------

def duration_category(row):

    if pd.isna(row['duration_value']):
        return 'Unknown'

    if row['type'] == 'Movie':
        if row['duration_value'] < 90:
            return 'Short'
        elif row['duration_value'] <= 120:
            return 'Medium'
        else:
            return 'Long'

    else:
        if row['duration_value'] <= 2:
            return 'Few Seasons'
        elif row['duration_value'] <= 5:
            return 'Medium Seasons'
        else:
            return 'Many Seasons'


df_features['duration_category'] = (
    df_features.apply(duration_category, axis=1)
)


# ------------------------------------------------------------
# 4. Create Content Added Year
# Useful for time-based analysis
# ------------------------------------------------------------

if 'added_year' not in df_features.columns:
    df_features['added_year'] = (
        pd.to_datetime(
            df_features['date_added'],
            errors='coerce'
        ).dt.year
    )


# ------------------------------------------------------------
# 5. Create Content Age Group
# Groups content based on release year
# ------------------------------------------------------------

def age_group(age):

    if pd.isna(age):
        return 'Unknown'

    if age <= 5:
        return '0-5 Years'
    elif age <= 10:
        return '6-10 Years'
    elif age <= 20:
        return '11-20 Years'
    else:
        return '20+ Years'


df_features['content_age_group'] = (
    df_features['content_age'].apply(age_group)
)


# ------------------------------------------------------------
# 6. Check correlation between numerical features
# ------------------------------------------------------------

numerical_features = df_features.select_dtypes(
    include=np.number
)

print("Correlation Matrix:")
display(numerical_features.corr())


# ------------------------------------------------------------
# 7. Display newly created features
# ------------------------------------------------------------

new_columns = [
    'content_age',
    'is_recent',
    'duration_category',
    'added_year',
    'content_age_group'
]

print("\nNewly Created Features:")
display(df_features[new_columns].head(10))


# ------------------------------------------------------------
# 8. Final dataset information
# ------------------------------------------------------------

print("\nFinal Dataset Shape:", df_features.shape)
print("\nFinal Columns:")
print(df_features.columns.tolist())
````

````python
# Handling Imbalanced Dataset (If needed)
# ============================================================
# HANDLING IMBALANCED DATASET - RANDOM OVERSAMPLING
# Amazon Prime TV Shows & Movies - EDA Project
# ============================================================

from imblearn.over_sampling import RandomOverSampler

# ------------------------------------------------------------
# 1. Check class distribution before oversampling
# ------------------------------------------------------------
print("Class Distribution Before Oversampling:")
print(y_train.value_counts())

# ------------------------------------------------------------
# 2. Apply Random Oversampling only on training data
# ------------------------------------------------------------
oversampler = RandomOverSampler(
    random_state=42
)

X_train_resampled, y_train_resampled = oversampler.fit_resample(
    X_train,
    y_train
)

# ------------------------------------------------------------
# 3. Check class distribution after oversampling
# ------------------------------------------------------------
print("\nClass Distribution After Oversampling:")
print(y_train_resampled.value_counts())

# ------------------------------------------------------------
# 4. Compare dataset sizes
# ------------------------------------------------------------
print("\nBefore Oversampling:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nAfter Oversampling:")
print("X_train_resampled:", X_train_resampled.shape)
print("y_train_resampled:", y_train_resampled.shape)

print("\nImbalanced Dataset Handling Completed Successfully!")
````
